# DRT planning toolkit walkthrough
Run from the toolkit folder. This notebook wraps the same verified CLI stages. The default config evaluates real historical Taiwan data and uses synthetic demand, network and proposed-service examples. Read docs/DRT_Planning_Guidebook.pdf before using your own inputs. Jupyter is an optional interface; the PowerShell/CLI workflow needs no Jupyter installation.

```python
from pathlib import Path
from datetime import datetime
import subprocess, sys
import pandas as pd

ROOT = Path.cwd()
assert (ROOT / "config/demo.json").exists(), "Open this notebook from the toolkit root"
CONFIG = "config/demo.json"  # change to config/my_project.json for your local study
RUN_ID = "notebook_" + datetime.now().strftime("%Y%m%d_%H%M%S")

def run(stage):
    subprocess.run([sys.executable, "-m", "drt_toolkit", stage, "--config", CONFIG, "--run-id", RUN_ID], cwd=ROOT, check=True)

print("Configuration:", CONFIG, "Run:", RUN_ID)
```

# DRT analysis toolkit walkthrough
The included data support 200 selected historical DMUs and all eight 200 m / 500 m fleet scenarios. All demographic and destination features are already prepared. No catchment analysis is performed. Use the same Python environment as requirements.txt and launch this notebook from the toolkit root.

```python
from pathlib import Path
from datetime import datetime
import subprocess, sys
import pandas as pd
ROOT = Path.cwd()
assert (ROOT / "config/case_study.json").exists(), "Open from the toolkit root"
CONFIG = "config/case_study.json"
RUN_ID = "notebook_" + datetime.now().strftime("%Y%m%d_%H%M%S")
def run(stage):
    subprocess.run([sys.executable, "-m", "drt_toolkit", stage, "--config", CONFIG, "--run-id", RUN_ID], check=True, cwd=ROOT)
print(CONFIG, RUN_ID)
```

Check packaged input paths and the restricted evaluation sample.
